```python
import cv2
import mediapipe as mp
import numpy as np
import matplotlib.pyplot as plt
import time

from dollarpy import Recognizer, Template, Point

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

print("All libraries imported successfully!")
```

```python
def extract_hand_position(hand_landmarks):
    landmarks = hand_landmarks.landmark

    # Wrist and palm landmarks
    ids = [0, 5, 9, 13, 17]

    x = np.mean([landmarks[i].x for i in ids])
    y = np.mean([landmarks[i].y for i in ids])

    return float(x), float(y)
```

```python
cap = cv2.VideoCapture(0)

with mp_hands.Hands(
    max_num_hands=1,
    min_detection_confidence=0.6,
    min_tracking_confidence=0.6
) as hands:

    while True:
        ret, frame = cap.read()

        if not ret:
            print("Camera error")
            break

        frame = cv2.flip(frame, 1)

        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = hands.process(rgb)

        if results.multi_hand_landmarks:
            hand = results.multi_hand_landmarks[0]

            mp_drawing.draw_landmarks(
                frame,
                hand,
                mp_hands.HAND_CONNECTIONS
            )

            x, y = extract_hand_position(hand)

            h, w = frame.shape[:2]

            cv2.circle(
                frame,
                (int(x * w), int(y * h)),
                10,
                (0, 255, 0),
                -1
            )

            cv2.putText(
                frame,
                "Hand Tracking",
                (20, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0, 255, 0),
                2
            )

        cv2.imshow("Bye Bye - Hand Tracking", frame)

        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

cap.release()
cv2.destroyAllWindows()
```

```python
# Cell 5 - Record Gesture Samples

gesture_samples = {
    "Bye Bye": [],
    "Other": []
}

RECORD_SECONDS = 3
NUM_SAMPLES = 5

cap = cv2.VideoCapture(0)

recording = False
current_points = []
current_label = None

with mp_hands.Hands(
    max_num_hands=1,
    min_detection_confidence=0.6,
    min_tracking_confidence=0.6
) as hands:

    while cap.isOpened():
        ret, frame = cap.read()

        if not ret:
            break

        frame = cv2.flip(frame, 1)
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = hands.process(rgb)

        if results.multi_hand_landmarks:
            hand = results.multi_hand_landmarks[0]

            mp_drawing.draw_landmarks(
                frame, hand, mp_hands.HAND_CONNECTIONS
            )

            if recording:
                x, y = extract_hand_position(hand)
                current_points.append((x, y))

        if recording:
            remaining = RECORD_SECONDS - (time.time() - start_time)

            cv2.putText(
                frame,
                f"Recording {current_label}: {max(0, remaining):.1f}s",
                (20, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 0, 255),
                2
            )

            if remaining <= 0:
                recording = False

                if len(current_points) >= 15:
                    gesture_samples[current_label].append(
                        current_points.copy()
                    )
                    print(
                        f"{current_label} saved: "
                        f"{len(gesture_samples[current_label])}/5"
                    )
                else:
                    print("Too few frames. Try again.")

        else:
            cv2.putText(
                frame,
                "B = Bye Bye | O = Other | Q = Quit",
                (20, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 255, 0),
                2
            )

        cv2.putText(
            frame,
            f"Bye Bye: {len(gesture_samples['Bye Bye'])}/5",
            (20, 90),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.65,
            (255, 255, 255),
            2
        )

        cv2.putText(
            frame,
            f"Other: {len(gesture_samples['Other'])}/5",
            (20, 125),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.65,
            (255, 255, 255),
            2
        )

        cv2.imshow("Gesture Recording", frame)

        key = cv2.waitKey(1) & 0xFF

        if not recording:
            if key == ord("b") and len(gesture_samples["Bye Bye"]) < 5:
                current_label = "Bye Bye"
                current_points = []
                start_time = time.time()
                recording = True

            elif key == ord("o") and len(gesture_samples["Other"]) < 5:
                current_label = "Other"
                current_points = []
                start_time = time.time()
                recording = True

        if key == ord("q"):
            break

        if all(
            len(v) >= NUM_SAMPLES
            for v in gesture_samples.values()
        ):
            break

cap.release()
cv2.destroyAllWindows()
```

```python
print("Bye Bye Samples:", len(gesture_samples["Bye Bye"]))
print("Other Samples:", len(gesture_samples["Other"]))

for label, samples in gesture_samples.items():
    print(f"\n{label}:")
    for i, sample in enumerate(samples):
        print(f"Sample {i+1}: {len(sample)} frames")
```

```python
# Cell 7 - Create DollarPy Templates

templates = []

for label, samples_list in gesture_samples.items():

    for sample in samples_list:

        points = [
            Point(float(x), float(y), 1)
            for x, y in sample
        ]

        templates.append(
            Template(label, points)
        )

print("Total Templates:", len(templates))
```

```python
# Cell 8 - Initialize Recognizer

recognizer = Recognizer(templates)

print("DollarPy Recognizer Ready!")
```

```python
# Cell 9 - Test Recorded Samples

for label, samples_list in gesture_samples.items():

    print(f"\nActual Gesture: {label}")

    for i, sample in enumerate(samples_list):

        points = [
            Point(float(x), float(y), 1)
            for x, y in sample
        ]

        name, score = recognizer.recognize(points)

        print(
            f"Sample {i+1}: "
            f"Predicted = {name}, "
            f"Score = {score:.4f}"
        )
```

```python
# Cell 10 - Live Gesture Recognition

cap = cv2.VideoCapture(0)

recording = False
live_points = []
result_text = "Press R to recognize"
RECORD_SECONDS = 3

with mp_hands.Hands(
    max_num_hands=1,
    min_detection_confidence=0.6,
    min_tracking_confidence=0.6
) as hands:

    while cap.isOpened():
        ret, frame = cap.read()

        if not ret:
            break

        frame = cv2.flip(frame, 1)
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = hands.process(rgb)

        if results.multi_hand_landmarks:
            hand = results.multi_hand_landmarks[0]

            mp_drawing.draw_landmarks(
                frame, hand, mp_hands.HAND_CONNECTIONS
            )

            x, y = extract_hand_position(hand)

            h, w = frame.shape[:2]
            cv2.circle(
                frame,
                (int(x * w), int(y * h)),
                10,
                (0, 255, 0),
                -1
            )

            if recording:
                live_points.append((x, y))

        if recording:
            remaining = RECORD_SECONDS - (time.time() - start_time)

            cv2.putText(
                frame,
                f"Recording: {max(0, remaining):.1f}s",
                (20, 90),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

            if remaining <= 0:
                recording = False

                if len(live_points) >= 15:
                    points = [
                        Point(float(x), float(y), 1)
                        for x, y in live_points
                    ]

                    name, score = recognizer.recognize(points)

                    result_text = f"{name}: {score:.3f}"

                    print(
                        f"Recognition Result: {name} "
                        f"| Score: {score:.4f}"
                    )
                else:
                    result_text = "Not enough frames"

        cv2.putText(
            frame,
            result_text,
            (20, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.75,
            (0, 255, 0),
            2
        )

        cv2.imshow("Live Bye Bye Recognition", frame)

        key = cv2.waitKey(1) & 0xFF

        if key == ord("r") and not recording:
            live_points = []
            start_time = time.time()
            recording = True
            result_text = "Recording..."

        if key == ord("q"):
            break

cap.release()
cv2.destroyAllWindows()
```
